## CS3: Effect Size

### Analysis:
The documentation demonstrates non-trivial effect sizes relative to baselines:

**Effect Size Evidence:**
1. **Correlation Analysis (Figure 2a):** Clear separation between max correlation distribution and baseline correlation distribution; excess correlations range up to ~0.8

2. **Universal vs Non-Universal Properties (Figure 3):** Universal neurons show dramatically different percentile rankings (80-100th percentile) vs non-universal neurons for weight norm, activation kurtosis, skew etc.

3. **Entropy Neurons (Figure 7):**
   - Layer norm scale changes from ~0.65 to ~0.80 with intervention (>20% change)
   - Entropy increases from ~2.4 to ~3.6 (>50% relative change)
   - Clear difference between entropy neurons and 20 random control neurons

4. **Prediction Neurons (Figure 6):** Kurtosis increases from ~4 to ~18 in later layers (4.5x increase)

5. **Attention Deactivation (Figure 8):** High correlations (ρ = 0.94, 0.97) between heuristic score and ablation effects

**Baseline Comparisons:**
- Random rotation baseline for correlation (Section 4.1)
- 20 random neurons for entropy neuron comparison (Figure 7)

**Verdict: PASS** - Effects are quantitatively large relative to baselines with clear magnitudes documented.

## CS4: Justification of Steps and Intermediate Conclusions

### Analysis:
The documentation provides explicit justifications for methodological choices:

**Justified Steps:**
1. **Why study neuron universality:** Section 1 explains motivation - universal features should be prioritized as more fundamental (citing Olah et al., Bau et al., Olsson et al.)

2. **Why use correlation threshold 0.5:** Section 4.1 (page 5): "While there is no principled threshold at which a neuron should be deemed universal" - acknowledges this is an arbitrary choice but provides context via baseline comparison

3. **Why excess correlation metric:** Section 4.1 explicitly justifies: to verify "the importance of the neuron basis" vs random rotated basis

4. **Why automated tests (Equation 4):** Section 4.3 explains - "developed several hundred automated tests to classify neurons using algorithmically generated labels"

5. **Why study weights (Section 5):** Explicitly states to understand "downstream effects" rather than just "inputs which cause activation"

6. **Why entropy neuron hypothesis:** Section 5.2 (page 9): Justified by observing neuron with "lowest variance logit effect" yet "high weight norm" - leading to hypothesis about layer norm modulation

**However, some design choices lack full justification:**
- Why 100 million tokens specifically?
- Why these particular automated test categories?

**Verdict: PASS** - Key design choices and intermediate conclusions have explicit justifications, though some minor choices are not fully explained. The core methodology is well-justified.

## CS5: Statistical Significance Reporting

### Analysis:
The documentation's approach to uncertainty and statistical measures:

**Uncertainty Reporting Present:**
1. **Figure 2b:** Shows max-max vs min-max correlation range for each neuron
2. **Section 4.1 (page 5):** Reports mean difference = 0.049 for all neurons and 0.105 for universal neurons
3. **Figure 6:** Shows percentile bands (5%, 10%, 50%, 90%, 95%) and "shaded area denotes range across all five models"
4. **Figure 3:** Reports statistics as percentiles compared to neurons in the same layer
5. **Figures 1c, 19:** Show "shaded area denotes standard deviation around the mean" for position neurons

**Statistical Measures Missing:**
- No p-values reported for any claims
- No confidence intervals for the 1-5% universality estimate
- No formal statistical tests comparing universal vs non-universal neuron properties
- Correlation coefficients (ρ) reported in Figure 8 and 26 but without uncertainty bounds

**Causal Intervention Results:**
- Figure 7 shows intervention results but no error bars or confidence intervals
- Path ablation results (Figure 8) show best fit correlations but no statistical significance tests

**Verdict: FAIL** - While some variability measures are shown (percentile bands, shaded regions for std), the key claims lack formal statistical significance testing. No p-values, no confidence intervals on main findings, no formal hypothesis tests. The uncertainty reporting is descriptive rather than inferential.
