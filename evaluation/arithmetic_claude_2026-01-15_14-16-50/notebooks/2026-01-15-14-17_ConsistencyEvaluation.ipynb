```python
# Let's get the detailed results from the cache to verify the claims in the plan
# We need to compare results across different methods and layers

import json
from collections import defaultdict

# Get results for different methods on key tasks
def get_all_results(dataset='word2vec', prefix='no_prefix'):
    cache_base = f'/net/scratch2/smallyan/arithmetic_eval/cache/parallelograms/{dataset}/{prefix}'
    methods = ['concept', 'token', 'all', 'raw']
    
    all_results = defaultdict(dict)
    
    for method in methods:
        method_path = os.path.join(cache_base, method)
        if os.path.exists(method_path):
            for task in os.listdir(method_path):
                task_path = os.path.join(method_path, task)
                if os.path.isdir(task_path):
                    all_results[(method, task)] = {}
                    for f in os.listdir(task_path):
                        if f.endswith('_results.json') and 'rank' not in f:
                            layer = int(f.split('layer')[1].split('_')[0])
                            with open(os.path.join(task_path, f), 'r') as file:
                                data = json.load(file)
                                all_results[(method, task)][layer] = data['nn_acc']
    
    return all_results

results_word2vec = get_all_results('word2vec', 'no_prefix')

# Print results for capital-common-countries
print("Capital-common-countries results:")
print("="*60)
for method in ['concept', 'token', 'all', 'raw']:
    key = (method, 'capital-common-countries')
    if key in results_word2vec:
        print(f"\n{method}:")
        for layer in sorted(results_word2vec[key].keys()):
            print(f"  Layer {layer}: {results_word2vec[key][layer]:.4f}")
```

```python
# Check random guessing baseline - calculate number of unique neighbors for each task
def count_neighbors(task, dataset='word2vec'):
    sep = ' ' if dataset == 'word2vec' else '\t'
    with open(f'{repo_path}/data/{dataset}/{task}.txt', 'r') as f:
        stuff = f.read()
    task_lines = [l for l in stuff.split('\n')[1:] if l != '']
    neighbors = set([w for l in task_lines for w in l.split(sep)])
    return len(neighbors), len(task_lines)

# Get random guessing baselines for key tasks
key_tasks = ['capital-common-countries', 'family', 'gram5-present-participle', 'gram7-past-tense']
print("Random guessing baselines:")
print("="*60)
for task in key_tasks:
    n_neighbors, n_examples = count_neighbors(task)
    random_chance = 1.0 / n_neighbors
    print(f"{task}:")
    print(f"  # unique words: {n_neighbors}")
    print(f"  # examples: {n_examples}")
    print(f"  Random chance: {random_chance:.4f} ({random_chance*100:.2f}%)")
```
