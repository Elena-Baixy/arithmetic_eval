```python
# Let me verify more results. The documentation says:
# - Capital cities: concept 80%, raw 47% at layer 20
# - Check all layers to find max performance for each method

import json
import os 

# Collect all results across layers
def collect_all_results(task, setting, dataset='word2vec', prefix='with_prefix'):
    results = {}
    base_path = f'/net/scratch2/smallyan/arithmetic_eval/cache/parallelograms/{dataset}/{prefix}/{setting}/{task}'
    if not os.path.exists(base_path):
        return results
    for f in os.listdir(base_path):
        if f.startswith('layer') and f.endswith('_results.json') and 'rank' not in f:
            layer = int(f.split('layer')[1].split('_')[0])
            with open(os.path.join(base_path, f), 'r') as fp:
                results[layer] = json.load(fp)['nn_acc']
    return results

print("Full layer analysis for key tasks:")
print("="*80)

tasks_to_check = ['capital-common-countries', 'family', 'gram5-present-participle', 'gram7-past-tense']

for task in tasks_to_check:
    print(f"\n{task}:")
    for setting in ['concept', 'token', 'raw', 'all']:
        results = collect_all_results(task, setting)
        if results:
            best_layer = max(results, key=results.get)
            best_acc = results[best_layer]
            print(f"  {setting:10s}: best={best_acc:.3f} at layer {best_layer}, all layers: {dict(sorted(results.items()))}")
```

```python
# Now let me also check the rank analysis data to verify the plan claims about effective rank
# Plan: "Performance maintained down to r=256"

def collect_rank_results(task, setting, layer, dataset='word2vec', prefix='with_prefix'):
    results = {}
    base_path = f'/net/scratch2/smallyan/arithmetic_eval/cache/parallelograms/{dataset}/{prefix}/{setting}/{task}'
    if not os.path.exists(base_path):
        return results
    for f in os.listdir(base_path):
        if f.startswith(f'layer{layer}') and f.endswith('_results.json'):
            if 'rank' in f:
                rank = int(f.split('rank')[1].split('_')[0])
                with open(os.path.join(base_path, f), 'r') as fp:
                    results[rank] = json.load(fp)['nn_acc']
            else:
                # Full rank (4096)
                with open(os.path.join(base_path, f), 'r') as fp:
                    results[4096] = json.load(fp)['nn_acc']
    return results

print("\nRank Analysis for Key Tasks:")
print("="*80)

# Check capital-common-countries at layer 20
print("\ncapital-common-countries (layer 20):")
for setting in ['concept', 'token', 'all']:
    results = collect_rank_results('capital-common-countries', setting, 20)
    if results:
        sorted_results = dict(sorted(results.items()))
        print(f"  {setting:10s}: {sorted_results}")

# Check gram7-past-tense at layer 16
print("\ngram7-past-tense (layer 16):")
for setting in ['concept', 'token', 'all']:
    results = collect_rank_results('gram7-past-tense', setting, 16)
    if results:
        sorted_results = dict(sorted(results.items()))
        print(f"  {setting:10s}: {sorted_results}")
```
