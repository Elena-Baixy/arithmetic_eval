```python
# Now let's properly load the cached results
import os
import json
import re

def load_cached_results(base_path, lens_type, task):
    """Load cached parallelogram results"""
    task_path = os.path.join(base_path, lens_type, task)
    results = {}
    if os.path.exists(task_path):
        for layer_file in os.listdir(task_path):
            if layer_file.endswith('.json'):
                # Parse layer number from filename like "layer20_results.json" or "layer20_rank128_results.json"
                match = re.match(r'layer(\d+)_results\.json', layer_file)
                if match:
                    layer = int(match.group(1))
                    with open(os.path.join(task_path, layer_file), 'r') as f:
                        results[layer] = json.load(f)
    return results

base_path = '/net/scratch2/smallyan/arithmetic_eval/cache/parallelograms/word2vec/with_prefix'

# Load results for the key tasks mentioned in replicated documentation
tasks_to_check = ['capital-common-countries', 'family', 'gram5-present-participle', 'gram7-past-tense']
lens_types = ['concept', 'token', 'all', 'raw']

print("=== CACHED RESULTS FROM ORIGINAL ===\n")

original_results = {}

for task in tasks_to_check:
    print(f"\n--- {task} ---")
    original_results[task] = {}
    for lens_type in lens_types:
        results = load_cached_results(base_path, lens_type, task)
        original_results[task][lens_type] = results
        if results:
            # Get results for layers 16 and 20
            for layer in [16, 20]:
                if layer in results:
                    data = results[layer]
                    if 'nn_correct' in data:
                        nn_correct = data['nn_correct']
                        total = len(nn_correct)
                        accuracy = sum(nn_correct) / total * 100
                        print(f"  {lens_type} L{layer}: {accuracy:.1f}%")
```

```python
# Great! Now let's get the nn_acc (nearest neighbor accuracy) for comparison
import os
import json
import re

def load_cached_results_v2(base_path, lens_type, task):
    """Load cached parallelogram results - get nn_acc from files"""
    task_path = os.path.join(base_path, lens_type, task)
    results = {}
    if os.path.exists(task_path):
        for layer_file in os.listdir(task_path):
            if layer_file.endswith('.json'):
                # Parse layer number from filename like "layer20_results.json"
                match = re.match(r'layer(\d+)_results\.json', layer_file)
                if match:
                    layer = int(match.group(1))
                    with open(os.path.join(task_path, layer_file), 'r') as f:
                        data = json.load(f)
                        if 'nn_acc' in data:
                            results[layer] = data['nn_acc'] * 100  # Convert to percentage
    return results

base_path = '/net/scratch2/smallyan/arithmetic_eval/cache/parallelograms/word2vec/with_prefix'

# Load results for the key tasks mentioned in replicated documentation
tasks_to_check = ['capital-common-countries', 'family', 'gram5-present-participle', 'gram7-past-tense']
lens_types = ['concept', 'token', 'all', 'raw']

print("=== CACHED RESULTS FROM ORIGINAL (NN Accuracy %) ===\n")

original_results = {}

for task in tasks_to_check:
    print(f"\n--- {task} ---")
    original_results[task] = {}
    for lens_type in lens_types:
        results = load_cached_results_v2(base_path, lens_type, task)
        original_results[task][lens_type] = results
        if results:
            # Get results for layers 16 and 20
            for layer in [16, 20]:
                if layer in results:
                    accuracy = results[layer]
                    print(f"  {lens_type} L{layer}: {accuracy:.1f}%")
```
