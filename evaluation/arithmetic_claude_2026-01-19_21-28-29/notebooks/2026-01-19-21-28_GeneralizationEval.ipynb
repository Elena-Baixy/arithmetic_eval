# Generalization Evaluation - Documentation Only

## Evaluation Mode: Documentation-Only Evaluation

This notebook evaluates the generalizability of findings reported in the documentation for "Vector Arithmetic in Concept and Token Subspaces" based **solely on explicit statements in the documentation**.

---

## Documentation Summary

The paper investigates how concept and token induction heads in Llama-2-7b can be used to identify subspaces of model activations that exhibit coherent semantic structure. Key findings:

1. **Concept lens** (using top-k concept induction heads' OV matrices) enables more accurate parallelogram arithmetic for semantic tasks (e.g., Athens - Greece + China = Beijing) - achieving ~80% accuracy vs 47% with raw hidden states.

2. **Token lens** (using top-k token induction heads' OV matrices) enables accurate arithmetic for surface-level tasks (e.g., code - coding + dancing = dance).

3. The method builds on prior work by Feucht et al. [2] on concept and token induction heads.

---

## GT1: Generalization to a New Model

### Evaluation Criteria
**PASS** — The documentation explicitly explains **why the identified mechanism or finding should transfer to a different model**, based on architectural properties, mechanistic reasoning, or stated invariances.

**FAIL** — Generalization to other models is asserted without mechanistic explanation, or not addressed at all.

### Documentation Evidence

The documentation focuses **exclusively on Llama-2-7b**. Relevant observations:

1. The paper states: "We show that these heads can be used to identify subspaces of model activations that exhibit coherent semantic structure in **Llama-2-7b**" (Abstract, p.1)

2. The method relies on "concept induction heads" and "token induction heads" identified in prior work [2] specifically for Llama-2-7b.

3. Section 2 references that the approach "possibly [works] because transformer representations are interchangeable in intermediate layers [3]" - but this is cited as an empirical observation, not a mechanistic argument for generalization to other models.

4. **No explicit discussion** is provided about:
   - Whether concept/token induction heads exist in other models
   - What architectural properties would be required for the mechanism to transfer
   - Whether the findings depend on Llama-2-7b's specific architecture or training

### Assessment

**GT1 = FAIL**

The documentation does not provide explicit mechanistic reasoning for why the findings would transfer to other models. The work is presented as an analysis of Llama-2-7b specifically, with no discussion of model generalization.

---

## GT2: Generalization to New Data

### Evaluation Criteria
**PASS** — The documentation explicitly states **assumptions or conditions** under which the finding should hold on new or unseen data, and explains why those assumptions are reasonable.

**FAIL** — Generalization to new data is asserted without explicit assumptions, justification, or scope.

### Documentation Evidence

1. The evaluation uses established datasets from Mikolov et al. [4] and Todd et al. [6] containing word tuples for analogy tasks (Table 1, Appendix A).

2. The paper evaluates across **14 word2vec tasks** and additional function vector tasks, covering diverse domains:
   - Semantic tasks: country-capitals, family relations, nationality-adjective
   - Surface-level tasks: present participle, past tense, plurals, adjective-to-adverb

3. Section 3.2 notes limitations: "this approach is less effective for tasks that seem simpler, like present participles of verbs" and "The failure of many of these tasks is unsurprising: some tasks are many-to-one relations that may not be represented as parallelograms" (Figure 5 caption).

4. **No explicit discussion** is provided about:
   - Conditions under which the findings would generalize to new word pairs not in the test sets
   - Assumptions about vocabulary coverage or word frequency
   - How performance might vary with out-of-distribution words

### Assessment

**GT2 = FAIL**

While the documentation evaluates on diverse tasks, it does not explicitly state the assumptions or conditions under which findings would generalize to new/unseen data. The paper reports empirical results without providing a principled argument for data generalization beyond the evaluated datasets.

---

## GT3: Method / Specificity Generalizability

### Evaluation Criteria
**PASS** — The documentation explicitly argues that the method can be applied to other similar tasks or settings, explaining what properties are required.

**FAIL** — Method generality is claimed without justification, or applicability conditions are unspecified.

**NA** — No new method is proposed.

### Documentation Evidence

The documentation **does propose a new method**: the "concept lens" and "token lens" transformations that combine attention head OV matrices to reveal semantic and surface-level structure in hidden states.

**Method Description (Section 2):**
- The method computes $L_{C_k} = \sum_{(l,h) \in C_k} O_{(l,h)} V_{(l,h)}$ to build transformations that extract concept or token information from hidden states.
- The method is presented as a general approach applicable to any set of identified attention heads.

**Evidence for method generalization:**
1. The paper mentions the approach is "repurposed" from prior work [2] which used it for visualization - showing at least some prior generalization.

2. Section 3.3 shows the method works with low-rank approximations (r=256), suggesting robustness.

3. **However**, no explicit discussion of:
   - What properties attention heads must have for this method to be applicable
   - Whether the method requires pre-identified concept/token induction heads
   - What makes a head suitable for concept vs token lens construction

### Assessment

**GT3 = FAIL**

While a new method (concept/token lens) is proposed, the documentation does not explicitly argue for its generalizability to other tasks or settings. The applicability conditions (e.g., what properties heads must have, how to identify suitable heads in new contexts) are not specified.

## Overall Generalizability Assessment

The documentation presents a focused empirical study of vector arithmetic in Llama-2-7b's activation subspaces. While the findings are interesting and well-supported within the experimental scope, **the documentation lacks explicit justification for generalizability** in all three dimensions:

1. **Model generalization**: The study is limited to a single model (Llama-2-7b) with no discussion of what architectural or mechanistic properties would enable transfer.

2. **Data generalization**: The evaluation covers diverse tasks but without explicitly stating the conditions under which findings would hold for new words or domains.

3. **Method generalization**: The concept/token lens approach is novel but its applicability requirements are not articulated.

This is characteristic of a workshop paper focused on demonstrating a phenomenon rather than establishing broad generalizability claims.
